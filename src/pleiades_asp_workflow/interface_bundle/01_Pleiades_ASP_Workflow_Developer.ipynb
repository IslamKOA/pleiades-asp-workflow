```python
from pathlib import Path
import importlib.util
import sys

EXPECTED_VERSION = "1.5.1"

# Always load the helper shipped beside this notebook. This prevents an old
# asp_utils2.py elsewhere on sys.path from silently driving a newer notebook.
helper_path = (Path.cwd() / "asp_utils2.py").resolve()
if helper_path.is_file():
    spec = importlib.util.spec_from_file_location("asp_utils2", helper_path)
    asp_utils2 = importlib.util.module_from_spec(spec)
    sys.modules["asp_utils2"] = asp_utils2
    spec.loader.exec_module(asp_utils2)
else:
    import asp_utils2

if getattr(asp_utils2, "DEV_VERSION", None) != EXPECTED_VERSION:
    raise RuntimeError(
        f"This notebook requires asp_utils2 v{EXPECTED_VERSION}, "
        f"but loaded v{getattr(asp_utils2, 'DEV_VERSION', 'unknown')} from "
        f"{getattr(asp_utils2, '__file__', 'unknown path')}. "
        "Use the asp_utils2.py supplied with this release."
    )

ui = asp_utils2.project_setup()
```
