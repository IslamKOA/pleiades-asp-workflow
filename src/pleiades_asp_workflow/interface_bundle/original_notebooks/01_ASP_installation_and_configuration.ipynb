# ASP Installation and Configuration

This notebook documents the installation and configuration of the Ames Stereo Pipeline (ASP) used in the Pléiades DSM generation workflow.

ASP is an open-source photogrammetric software package used for stereo image processing, bundle adjustment, map projection, point-cloud generation, and DEM/DSM production. In this workflow, ASP tools such as `bundle_adjust`, `mapproject`, `parallel_stereo`, and `point2dem` are required for processing raw Pléiades tri-stereo imagery.

For reproducibility, this workflow is standardized using **ASP version 3.3.0**.

## Step 1: Download and Extract ASP

ASP can be downloaded from the official GitHub release page:

https://github.com/NeoGeographyToolkit/StereoPipeline/releases

In this work, the processing workflow was performed using **ASP 3.3.0**. Although newer ASP versions are available, ASP 3.3.0 was retained after workflow testing because it provided the most stable and consistent results for our Pléiades sensor-model refinement and bundle-adjustment experiments.

After downloading the Linux release, place the archive in a fixed installation directory, for example:

```bash
/mnt/summer/USERS/KOAI/ASP

````python
%%bash
# ------------------------------------------------------------
# Step 1: Define ASP installation paths
# ------------------------------------------------------------

ASP_DIR="/mnt/summer/USERS/KOAI/ASP"
ASP_ARCHIVE="StereoPipeline-3.3.0-Linux.tar.bz2"
ASP_FOLDER="StereoPipeline-3.3.0-Linux"

echo "ASP installation directory:"
echo "$ASP_DIR"

echo ""
echo "Expected ASP archive:"
echo "$ASP_ARCHIVE"

echo ""
echo "Expected ASP folder after extraction:"
echo "$ASP_FOLDER"
````

````python
%%bash
# ------------------------------------------------------------
# Step 2: Extract ASP archive
# ------------------------------------------------------------

ASP_DIR="/mnt/summer/USERS/KOAI/ASP"
ASP_ARCHIVE="StereoPipeline-3.3.0-Linux.tar.bz2"
ASP_FOLDER="StereoPipeline-3.3.0-Linux"

# Go to the ASP installation directory
cd "$ASP_DIR" || {
    echo "ERROR: ASP directory does not exist: $ASP_DIR"
    exit 1
}

# Check that the ASP archive exists
if [ ! -f "$ASP_ARCHIVE" ]; then
    echo "ERROR: ASP archive not found:"
    echo "$ASP_DIR/$ASP_ARCHIVE"
    exit 1
fi

# Extract ASP only if the folder does not already exist
if [ ! -d "$ASP_FOLDER" ]; then
    echo "Extracting ASP archive..."
    tar -xvf "$ASP_ARCHIVE"
else
    echo "ASP folder already exists:"
    echo "$ASP_DIR/$ASP_FOLDER"
fi
````

````python
%%bash
# ------------------------------------------------------------
# Step 3: Check ASP binary directory
# ------------------------------------------------------------

ASP_DIR="/mnt/summer/USERS/KOAI/ASP"
ASP_FOLDER="StereoPipeline-3.3.0-Linux"
ASP_BIN="$ASP_DIR/$ASP_FOLDER/bin"

if [ ! -d "$ASP_BIN" ]; then
    echo "ERROR: ASP bin directory not found:"
    echo "$ASP_BIN"
    exit 1
fi

echo "ASP bin directory found:"
echo "$ASP_BIN"

echo ""
echo "Available ASP binaries:"
ls -lh "$ASP_BIN" | head -n 20
````

## Step 2: Prepare the Python Environment with Micromamba

Before using ASP inside Jupyter, it is recommended to create a clean Python environment.

This part should be done from a real Linux terminal, not from inside Jupyter.

We use `micromamba` because it is lightweight and useful for managing isolated Python environments. In this workflow, the environment is called:

```bash
asp

### 2.1 Install Micromamba

Open a Linux terminal and run:

```bash
"${SHELL}" <(curl -L micro.mamba.pm/install.sh)

After installation, close and reopen the terminal, or run:
source ~/.bashrc

Then check that micromamba is available:
micromamba --version

### 2.2 Create the ASP Python Environment

Create a new environment called `asp`:

```bash
micromamba create -n asp -c conda-forge python=3.10 jupyterlab notebook ipykernel numpy pandas matplotlib rasterio geopandas scipy -y
```

Activate the environment:

```bash
micromamba activate asp
```

Register the environment as a Jupyter kernel:

```bash
python -m ipykernel install --user --name asp --display-name "Python (asp)"
```

This allows the notebook to use the correct Python environment.

## Step 3: Add ASP to the System PATH

After creating and activating the `asp` environment, add the ASP `bin` directory to your shell configuration.

Open `.bashrc`:

```bash
nano ~/.bashrc
```

Add this line at the end of the file:

```bash

```

Save and close:

```text

Reload `.bashrc`:

```

Then check that ASP is available:
