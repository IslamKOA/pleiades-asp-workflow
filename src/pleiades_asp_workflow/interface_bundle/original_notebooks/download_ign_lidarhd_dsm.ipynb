# Download and prepare the IGN LiDAR HD DSM

This notebook downloads the **IGN LiDAR HD MNS**, corresponding to a digital surface model (DSM), for a user-defined area of interest. The native 0.5 m GeoTIFF tiles are selected from the IGN WFS tile index, downloaded, checked, aggregated to the requested resolution, mosaicked, and cropped.

The detailed functions are stored in `ign_lidarhd_downloader.py` so that the notebook remains concise and reproducible.

> This workflow downloads the derived MNS/DSM raster product. It does not download or process the raw LiDAR point clouds.

## Import the workflow

```python
from pathlib import Path

from ign_lidarhd_downloader import download_ign_lidarhd_dsm
```

## User settings

Only the AOI path and output directory normally need to be changed. The remaining values reproduce the settings of the original R workflow.

```python
AOI_PATH = Path("/mnt/summer/USERS/KOAI/Cosmo-SkyMed/shapefile/AOI.shp")

# Use a clean output directory to avoid mixing files from previous tests.
OUT_DIR = Path("/mnt/summer/USERS/KOAI/Cosmo-SkyMed/reference_DEMs/IGN_LiDARHD_DSM_MNS")

TARGET_RES = 50

DOWNLOAD_BUFFER = True
BUFFER_DISTANCE_M = 1000

N_WORKERS = 2

EXPORT_TO_WGS84 = True
OUTPUT_CRS = "EPSG:4326"
WGS84_PROJECT_METHOD = "nearest"

PLOT_RESULTS = True
```

## Run the complete workflow

```python
outputs = download_ign_lidarhd_dsm(
    aoi_path=AOI_PATH,
    out_dir=OUT_DIR,
    target_res=TARGET_RES,
    download_buffer=DOWNLOAD_BUFFER,
    buffer_distance_m=BUFFER_DISTANCE_M,
    n_workers=N_WORKERS,
    aggregation_fun="mean",
    export_to_wgs84=EXPORT_TO_WGS84,
    output_crs=OUTPUT_CRS,
    wgs84_project_method=WGS84_PROJECT_METHOD,
    plot_results=PLOT_RESULTS,
    max_download_attempts=3,
)
```

## Output paths

```python
for name, path in outputs.items():
    print(f"{name}: {path}")
```

## Main outputs

The workflow produces:

- Native 0.5 m MNS tiles downloaded from IGN.
- Individual tiles aggregated to the target resolution.
- A bounding-box DSM in EPSG:2154.
- An AOI-masked DSM in EPSG:2154.
- NoData diagnostic rasters.
- Per-tile NoData statistics.
- Optional EPSG:4326 copies.
- A GeoPackage containing the selected IGN tile index.

## Data source

IGN LiDAR HD MNS tiles are distributed as 1 km × 1 km GeoTIFF files at 0.5 m spacing. Availability depends on the current national production coverage, and restricted areas may contain NoData values.

- IGN LiDAR HD information: https://geoservices.ign.fr/lidarhd
- MNS LiDAR HD dataset: https://www.data.gouv.fr/datasets/mns-lidar-hd
- IGN Géoplateforme WFS: https://data.geopf.fr/wfs/ows?SERVICE=WFS&REQUEST=GetCapabilities&VERSION=2.0.0
